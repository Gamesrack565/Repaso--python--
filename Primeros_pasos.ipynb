## Listas (Manejo de colecciones)
Una lista es una estructura de datos que python interpreta como una colección ordenada de elementos

Cada elemento se almacena en una posicion:



```python
lista: ["manzana", "pera", "uva"]
#indice:   0          1       2
```

#### Mutabilidad
Las listas si se pueden modificar

Se pueden cambiar, agregar o eliminar elementos sin crear una nueva lista

#### Indexado y slicing
Permite acceder a un elemento especifico o extraer una parte de la lista sin modificar la original

#### Agregar elementos
Permite incluir nuevos datos al final o en una posicion determinada

#### Eliminar elementos
Permite quitar elementos por valor, por indice o usando palabras clave

#### Métodos comunes
Las listas tienen metodos utiles como append, insert, remove y pop para trabajar con sus elementos.

```python
lista = ["elemento1", "elemento2", "elemento3"]
print(f"Lista original: {lista}")

#Modificacion de una lista
lista[1] = "Agua"

#Agregar nuevos elementos al final
lista.append("mayo")

#Inseratmos un elemento en cierta ubicacion
lista.insert(2, 21)

print(f"Lista modificada: {lista}\n")
```

#### Eliminar elementos de una lista
Permite borrar elementos de forma directa segun el valor o la posicion

```python
#Metodo remove: (Se debe indicar el elemento que se eliminara)
print(f"Lista original: {lista}")
print("---> lista.remove(valor)")
lista.remove(21)
print("Valor = 21")
print(f"Lista modificada: {lista}\n")

#Metodo POP: remueve (saca) por indice
print(f"Lista original: {lista}")
print("---> lista.pop(indice)")
lista.pop(1)
print("Indice = 1")
print(f"Lista modificada: {lista}\n")

#Metodo del: Elimina indice
print(f"Lista original: {lista}")
print("---> del lista[indice]")
del lista[2]
print("Indice = 2")
print(f"Lista modificada: {lista}\n")
```

#### Obtener sublistas
Permite extraer una parte de la lista original sin modificarla

```python
## Obtener sublistas
print("Obtener sublistas: (Indica donde empieza y en donde termina, sin contar el ultimo)")
lista = ["elemento1", "elemento2", "elemento3"]
print(f"Lista original: {lista}")
print("---> lista[donde, final-1]")
sublista = lista[1:3]
print(f"Sublista: {sublista}")
```

### Sets
Colecciones de ddatos no ordendas de elementos unicos. UTILES PARA EVITAR ELEMENTOS DUPLICADOS

No existen indices, por lo que no se puede acceder a un elemento en particular. 
    
    Se definen usando llaves {} o la función set().

```python
set1 = {1, 2, 3, 4, 5, 5, 5, 5, 5}
set2 = {77, 88, 99, 99, 5}

print(f"Set1 (No duplicados): {set1}") 

#Operaciones con sets:
print("\nOperaciones con sets:\n")
print("---> Union: a|b")
print(set1 | set2)

print("---> Interseccion: a&b")
print(set1 & set2)

print("---> Diferenica: a - b")
print(set1 - set2)
```

### Diccionarios
Son colecciones de datos no ordenadas que almacenan pares clave-valor. Cada elemento se accede mediante su clave única, en lugar de un índice numérico. 

    Se definen usando llaves {} y los elementos se separan con comas, con la sintaxis: {clave: valor}.

```python
diccionario = {'nombre': 'Angel', 'edad': 22}

print(diccionario)

#Acceder a una llave
print("\nMetodo 1: get")
print("---> diccionario.get(\"llave\")")
print(f"Edad = {diccionario.get('edad')}")

print("Metodo 2: get")
print("---> diccionario[\"llave\"]")
print(f"Edad = {diccionario['edad']}")
```

### Clases y objetos
Una clase es el "plano" o la plantilla. Se usa la palabra reservada class. 
El método __init__ es el constructor, que se ejecuta automáticamente cuando creas un nuevo objeto.


Una vez definida la clase, puedes crear tantos objetos (instancias) como quiera:

    objeto = NombreDeLaClase(parametros)

```python
class Persona:
    # Constructor: inicializa los atributos del objeto
    def __init__(self, nombre, edad):
        self.nombre = nombre  # Atributo
        self.edad = edad      # Atributo

    # Un método: una función que pertenece a la clase
    def saludar(self):
        print(f"Hola, mi nombre es {self.nombre} y tengo {self.edad} años.")

# Crear objetos
persona1 = Persona("Ana", 25)
persona2 = Persona("Carlos", 30)

# Usar los objetos
persona1.saludar()  # Salida: Hola, mi nombre es Ana y tengo 25 años.
print(persona2.nombre) # Salida: Carlos
```

Dirección de Memoria: Cuando creas un objeto, Python le asigna un espacio único en la memoria de tu computadora. Aunque dos objetos tengan exactamente los mismos datos, son entidades diferentes en memoria. Puedes verificar esto usando la función id().

Atributos Dinámicos: Python es tan flexible que te permite agregarle nuevos atributos a un objeto después de haberlo creado, incluso si ese atributo no se definió en la clase original. Sin embargo, este atributo solo existirá para ese objeto en específico.

```python
persona1 = Persona("Ana", 25)
persona2 = Persona("Ana", 25)

# Direcciones de memoria diferentes
print(id(persona1)) # Ej: 140321456...
print(id(persona2)) # Ej: 140321789... (Diferente)

# Atributo dinámico (solo se añade a persona1)
persona1.telefono = "555-1234"
print(persona1.telefono)
# print(persona2.telefono) -> Daría error, persona2 no tiene este atributo
```

### Encapsulamiento, Getters y Setters:
El encapsulamiento es un principio de la programación orientada a objetos que restringe el acceso directo a ciertos componentes de un objeto. Esto se hace para proteger los datos internos y garantizar que se acceda a ellos de manera controlada.

Getters y Setters son métodos que permiten obtener (getter) y modificar (setter) los valores de los atributos de un objeto, respetando el encapsulamiento.

Mejora de Encapsulamiento: Se utilizan los decoradores @property para el método get y @atributo.setter para el método set. Esto permite usar los métodos como si fueran atributos normales.

```python
class CuentaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular
        self._saldo = saldo # Atributo "privado" (protegido)

    # Método GET
    @property
    def saldo(self):
        return self._saldo

    # Método SET
    @saldo.setter
    def saldo(self, cantidad):
        if cantidad >= 0:
            self._saldo = cantidad
        else:
            print("El saldo no puede ser negativo")

# Uso
mi_cuenta = CuentaBancaria("Carlos", 1000)
print(mi_cuenta.saldo) # Llama al GET: 1000

mi_cuenta.saldo = 1500 # Llama al SET
print(mi_cuenta.saldo) # 1500

mi_cuenta.saldo = -500 # Llama al SET con validación
# Imprime: El saldo no puede ser negativo
```

### Atributos y metodos de clase 
Hasta ahora, los atributos y métodos pertenecían a los objetos (instancias). Sin embargo, también pueden pertenecer a la clase en sí misma.

Atributos de Clase: Son variables que se comparten entre todos los objetos de la misma clase. Si un objeto cambia ese atributo, cambia para todos.

Métodos de Clase: Se decoran con @classmethod y reciben la clase misma como primer argumento (cls), no la instancia (self). Se usan para interactuar con los atributos de clase.


#### Notas:
    @classmethod convierte un método en un método de clase. Eso significa que el primer parámetro no es self, sino cls, que representa a la clase completa, no a una instancia específica.

    cls no es una palabra reservada de Python, es solo el nombre por convención. Podrías poner otro nombre, pero cls se usa porque deja claro que ahí recibes la clase. Con eso puedes acceder a atributos de clase como contador_alumnos.

```python
class Alumnos:
    contador_alumnos = 0
    def __init__(self, nombre):
        self.alumno = nombre
        Alumnos.contador_alumnos += 1
        self.id = self.contador_alumnos

    @classmethod
    def total_alumnos(cls):
        print(f"Total de alumnso registrados: {cls.contador_alumnos}")

alumno1 = Alumnos("Robert")
print(f"Alumno 1: {alumno1.alumno}, ID: {alumno1.id}")
Alumnos.total_alumnos()
```
